```python
import numpy as np

from scipy.ndimage import gaussian_filter1d

import matplotlib.pyplot as plt

from matplotlib.animation import FuncAnimation, PillowWriter
```

```python
def x_derivative(f, h):
    dfdx = np.zeros_like(f)
    dfdx[1:-1, :] = (f[2:, :] - f[:-2, :]) / (2*h)
    return dfdx

def y_derivative(f, h):
    dfdy = np.zeros_like(f)
    dfdy[:, 1:-1] = (f[:, 2:] - f[:, :-2]) / (2*h)
    return dfdy

def laplacian(f, h):
    lap = np.zeros_like(f)
    lap[1:-1, 1:-1] = (f[2:, 1:-1] + f[:-2, 1:-1] + f[1:-1, 2:] + f[1:-1, :-2] - 4*f[1:-1, 1:-1]) / h**2
    return lap

def generate_double_well_gradient_X(X, a, b):
    fx = a*X - b*X**3
    return fx

def generate_double_well_gradient_Y(Y, c):
    fy = -Y * c
    return fy

def dpdt_array_generator(P_array, fx_array, fy_array, alpha, h):
    # Drift term: -d/dx[f_x P] - d/dy[f_y P]
    dfxP_dx = x_derivative(fx_array * P_array, h)
    dfyP_dy = y_derivative(fy_array * P_array, h)
    drift_term = -(dfxP_dx + dfyP_dy)

    # Diffusion term: alpha * Laplacian(P)
    diffusion_term = alpha * laplacian(P_array, h)

    return drift_term + diffusion_term
```

```python
# Domain (same spacing h in x and y)
h = 0.05
x_array = np.arange(-2, 2 + h/2, h)
y_array = np.arange(-2, 2 + h/2, h)

X, Y = np.meshgrid(x_array, y_array, indexing="ij")
```

```python
# Starting condition: 2D Gaussian centered at (x0, y0)
x0, y0 = 0, 0
sigma = 0.2
P_array = np.exp(-((X - x0)**2 + (Y - y0)**2) / (2*sigma**2))
P_array = P_array / (np.sum(P_array) * h**2)   # total probability = 1
```

```python
# Forces on the grid

a = 1
b = 2
c = 1
alpha = 0.25

fx_array = generate_double_well_gradient_X(X, a, b)
fy_array = generate_double_well_gradient_Y(Y, c)

```

```python
dt = 0.0004
t_end = 4
n_steps = int(t_end / dt)

#Saving graphs
n_saves = 50
save_every = n_steps // n_saves
P_history = [P_array]
t_history = [0]

for step in range(n_steps):
    dpdt_array = dpdt_array_generator(P_array, fx_array, fy_array, alpha, h)
    #Calculating the new P_array
    P_new = P_array + dpdt_array * dt
    P_array = P_new

    if (step + 1) % save_every == 0:
        P_history.append(P_array)
        t_history.append((step + 1) * dt)

P_fd, t_fd, x_fd, y_fd = np.array(P_history), np.array(t_history), x_array, y_array
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection="3d")
z_max = 0.5 * np.max(P_history)

def update(i):
    ax.clear()   # a 3D surface can't be updated in place, so redraw each frame
    ax.plot_surface(X, Y, P_history[i], cmap="viridis",
                rcount=50, ccount=50, linewidth=0, antialiased=False)
    ax.set_zlim(0, z_max)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("P")
    ax.set_title(f"t = {t_history[i]:.2f}")

anim = FuncAnimation(fig, update, frames=len(P_history), interval=50)
anim.save("fokker_planck_2d.gif", writer=PillowWriter(fps=20), dpi=80)
```

```python
# Asked AI to add a dimention to my old code, it is a simple deviation and I checked it over


import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter


# Gradient of V(x, y) = -a x^2/2 + b x^4/4 + c y^2/2
def gradV(x, y, a, b, c):
    dVdx = -a * x + b * x**3
    dVdy = c * y
    return dVdx, dVdy

def langevin_step(x, y, dt, noise_coef, rng, a, b, c):
    dVdx, dVdy = gradV(x, y, a, b, c)
    x_new = x - dt * dVdx + noise_coef * rng.standard_normal(len(x))
    y_new = y - dt * dVdy + noise_coef * rng.standard_normal(len(y))
    return x_new, y_new

a, b, c = 1, 2, 1
temp = 0.25            # same as alpha in the FD code
dt = 0.01
t_max = 4
n_particles = 1000000

rng = np.random.default_rng(42)
noise_coef = np.sqrt(2.0 * temp * dt)
n_steps = int(t_max / dt)

# Same starting Gaussian as the FD code
x_pos, y_pos = 0, 0
sigma = 0.2
x = x_pos + sigma * rng.standard_normal(n_particles)
y = y_pos + sigma * rng.standard_normal(n_particles)

n_saves = 100
save_every = n_steps // n_saves
x_history = [x]
y_history = [y]
t_history = [0]

for i in range(n_steps):
    x, y = langevin_step(x, y, dt, noise_coef, rng, a, b, c)

    if (i + 1) % save_every == 0:
        x_history.append(x)
        y_history.append(y)
        t_history.append((i + 1) * dt)

t_history = np.array(t_history)

# Bin edges: 50 bins from -2 to 2 in both x and y
bins = np.linspace(-2, 2, 101)

P_history = []
for k in range(len(x_history)):
    # Positions of all particles at save number k
    x_positions = x_history[k]
    y_positions = y_history[k]

    # Count how many particles fall in each 2D bin.
    # density=True turns the counts into a probability density (total probability = 1)
    histogram_result = np.histogram2d(x_positions, y_positions, bins=[bins, bins], density=True)

    # np.histogram2d returns three things: the densities, the x edges and the y edges.
    # We only need the densities.
    P = histogram_result[0]

    P_history.append(P)

P_history = np.array(P_history)
```

```python
# Plot as a color map
fig, ax = plt.subplots()

img = ax.imshow(P_history[0].T, origin="lower", cmap="viridis",
                extent=[-2, 2, -2, 2], vmin=0, vmax=0.6,
                interpolation="bilinear")
fig.colorbar(img, label="P")
ax.set_xlabel("x")
ax.set_ylabel("y")

def update(k):
    img.set_data(P_history[k].T)
    ax.set_title(f"t = {t_history[k]:.2f}")

anim = FuncAnimation(fig, update, frames=len(P_history), interval=50)
anim.save("monte_carlo_2d.gif", writer=PillowWriter(fps=20))
```

```python
### Just  to serve as confirmation


from scipy.interpolate import RegularGridInterpolator, interp1d

# Centers of the MC histogram bins, as a 2D grid
centers = (bins[:-1] + bins[1:]) / 2
CX, CY = np.meshgrid(centers, centers, indexing="ij")
points = np.stack([CX.ravel(), CY.ravel()], axis=-1)


### FOllowing are optimized AI code for binning the fokker planck same as the MC
# FD solution moved onto the MC bin centers (space)
P_fd_bins = np.array([ RegularGridInterpolator((x_fd, y_fd), P)(points).reshape(CX.shape) for P in P_fd])

# FD solution moved onto the MC save times (time)
P_fd_matched = interp1d(t_fd, P_fd_bins, axis=0, bounds_error=False,
                        fill_value=(P_fd_bins[0], P_fd_bins[-1]))(t_history)

# Exact stationary solution on the bin centers
V = -a*CX**2/2 + b*CX**4/4 + c*CY**2/2
P_exact = np.exp(-V / temp)
P_exact = P_exact / (np.sum(P_exact) * (bins[1] - bins[0])**2)

# RMS differences over all bins, at each time
rms_mc_fd = np.sqrt(np.mean((P_history - P_fd_matched)**2, axis=(1, 2)))
rms_fd_exact = np.sqrt(np.mean((P_fd_matched - P_exact)**2, axis=(1, 2)))
rms_mc_exact = np.sqrt(np.mean((P_history - P_exact)**2, axis=(1, 2)))

plt.figure()
plt.plot(t_history, rms_mc_fd, label="MC vs FD")
plt.plot(t_history, rms_fd_exact, label="FD vs exact")
plt.plot(t_history, rms_mc_exact, label="MC vs exact")
plt.xlabel("t")
plt.ylabel("RMS difference")
plt.legend()
plt.show()
```

```python
#AI Code
import os
from scipy.ndimage import gaussian_filter
from matplotlib.colors import LightSource

n_frames = 6                    # number of png frames per simulation
t_last = 4                      # time of the last frame
z_max = 1.5 * np.max(P_fd[-1])  # top of the P-axis, same for both so they can be compared
smooth_sigma = 1.5              # smoothing of the MC histogram, in bins (0 = no smoothing)
out_dir = "frames_2d"
os.makedirs(out_dir, exist_ok=True)

# Grids for plotting
X_fd, Y_fd = np.meshgrid(x_fd, y_fd, indexing="ij")
centers = (bins[:-1] + bins[1:]) / 2
X_mc, Y_mc = np.meshgrid(centers, centers, indexing="ij")

# n evenly spaced times from 0 to t_last
t_frames = np.linspace(0, t_last, n_frames)

# Light from the top left, gives the surface a shaded 3D look
light = LightSource(azdeg=315, altdeg=45)


def save_surface(Xg, Yg, P, t, title, filename):
    P_plot = np.minimum(P, z_max)    # cut off anything above the top of the plot
    colors = light.shade(P_plot, cmap=plt.cm.viridis, vmin=0, vmax=z_max,
                         vert_exag=0.5, blend_mode="soft")

    fig = plt.figure(figsize=(7, 5.5))
    ax = fig.add_subplot(projection="3d")
    ax.plot_surface(Xg, Yg, P_plot, facecolors=colors,
                    rcount=Xg.shape[0], ccount=Xg.shape[1],
                    linewidth=0, antialiased=False, shade=False)
    ax.set_proj_type("ortho")       # no perspective, so equal heights look equal
    ax.set_zlim(0, z_max)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_zlabel("P")
    ax.set_title(f"{title}, t = {t:.2f}")
    ax.view_init(elev=30, azim=-75)
    fig.savefig(filename, dpi=150, bbox_inches="tight")
    plt.close(fig)
    print("saved", filename)


for k, t in enumerate(t_frames):
    # Saved frame of each simulation closest to this time
    i_fd = np.argmin(np.abs(t_fd - t))
    i_mc = np.argmin(np.abs(t_history - t))

    # Fokker-Planck
    save_surface(X_fd, Y_fd, P_fd[i_fd], t_fd[i_fd], "Fokker-Planck",
                 os.path.join(out_dir, f"fp_frame_{k:02d}.png"))

    # Monte Carlo (histogram smoothed so the surface is not jagged)
    P_mc_smooth = gaussian_filter(P_history[i_mc], sigma=smooth_sigma)
    save_surface(X_mc, Y_mc, P_mc_smooth, t_history[i_mc], "Monte Carlo",
                 os.path.join(out_dir, f"mc_frame_{k:02d}.png"))
```
